```python
from rdkit import Chem
from rdkit.Chem import Descriptors
```

```python
smiles = "CC(=O)OC1=CC=CC=C1C(=O)O"

mol = Chem.MolFromSmiles(smiles)

print("Number of atoms:", mol.GetNumAtoms())
```

```python
for atom in mol.GetAtoms():
    print(
        "Index:",
        atom.GetIdx(),
        "| Element:",
        atom.GetSymbol(),
        "| Aromatic:",
        atom.GetIsAromatic()
    )
```
